## 2. The question: decision, action, cost of a wrong call

### Research question

Which pages should be reviewed first for a possible content refresh or other content action?

### Unit of analysis

The unit of analysis is one content page.

### Decision

The decision is which pages the content or SEO team should review first.

### Action

A reviewer could investigate the highest-ranked pages and decide whether to refresh, expand, protect, prune, or monitor them.

### Cost of a wrong recommendation

A wrong recommendation could waste the team's limited time and resources on pages that do not need attention. It could also cause an important page to be overlooked.

### Why data or ML can help

There may be patterns in search performance, traffic, content age, ranking position, CTR, and engagement that are difficult to identify by manually checking every page. Data or ML could help rank pages so that human reviewers can focus their attention on the most promising opportunities.
