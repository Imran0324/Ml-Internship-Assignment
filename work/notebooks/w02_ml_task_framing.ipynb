## 1. My lane as an ML task (type)

I am framing this as a **Ranking/Scoring** task. We want to score existing content by its likelihood of traffic decay so the content team can prioritize their refresh efforts. By assigning a score to each piece of content and sorting them, we effectively turn this into a ranking problem.

```python
# Code to check the different types of trend directions available
import pandas as pd
df = pd.read_csv("/content/Ml-Internship-Assignment/data/raw/content_refresh_anonymized.csv")
df['trend_direction'].value_counts(dropna=False)
```

## 2. Target or proxy

**Target**: The target we want to predict is `trend_direction` (specifically, whether the trend is 'down').
This label comes from an **observed outcome**: the actual shift in traffic metrics (like sessions or clicks) comparing the last 30 days to the previous 30 days, which the dataset already encodes as the `trend_direction`.

```python
# Sketching the target column
df[['content_id', 'trend_direction', 'trend_pct']].head()
```

## 3. Success metric

**Success metric**: Precision@K (Precision at K).
If the model recommends 100 articles to be refreshed (K=100), we measure what percentage of them were actually experiencing severe decay. A Precision@100 of > 80% would be a 'good' directional metric, meaning the content team's time is spent highly effectively.

```python
# Example of checking how many articles are actually going 'down' to see if our metric is viable
down_pct = (df['trend_direction'] == 'down').mean() * 100
print(f"{down_pct:.1f}% of the content is trending down. Random guessing would give a Precision of {down_pct:.1f}%")
```

## 4. The unit of analysis, as a real dataframe

**Unit of Analysis**: One row = **one piece of published content** (e.g., an article or page) belonging to a specific client.
Each row contains its historical performance, engagement metrics, and content characteristics.

```python
# Showing the unit of analysis as a dataframe
display(df.head(3))
print(f"Unit of analysis: 1 row = 1 piece of content. Total rows: {len(df)}")
```

## 5. Why ML beats a fixed rule here

**Why ML beats a fixed rule**:
A simple rule like "refresh if `content_age_days` > 365 and `clicks_last_30d` < 10" is too rigid. Traffic decay is messy and depends on non-linear combinations of factors. For instance, high-competition commercial articles might decay much faster than low-competition informational articles. Additionally, subtle engagement signals like `scroll_rate` or `engagement_rate` might be early indicators of decay before traffic fully drops. ML can learn these complex, multivariate interactions that a basic if-statement would miss.

```python
# Exploring the messiness of the pattern
df.groupby('content_type')['trend_pct'].mean()
```
